# Build, draw, and run one Program

This notebook prepares a Bell pair, draws its circuit, runs 1,000 shots, and plots the outcomes. Run each cell with **Shift+Enter**. The first cell takes longer while FatQat is imported.

```python
import fatqat as fq
import fatqat.operations as ops

program = fq.Program(2, 2)
program.add(ops.H, 0)
program.add(ops.CX, (0, 1))
program.measure_all()
```

`Program(2, 2)` creates two qubits and two classical bits. `H` puts qubit 0 into a superposition, `CX` entangles it with qubit 1, and the measurement writes both outcomes to the classical bits. Drawing shows the recorded instructions without running them:

```python
figure = program.draw()
```

Run the Program on the general-purpose simulator. In the browser, choose `runtime="numpy"`: the default numba runtime is not available here.

```python
backend = fq.simulator.Simulator(runtime="numpy")
result = backend.run(program, shots=1000, simulation_config={"seed": 7}).result()
result.get_counts()
```

Only `00` and `11` appear: the measured bits agree because the qubits are entangled. `Result.draw()` plots the counts:

```python
result.draw(title="Bell-state outcomes")
```
